# 02 · Factor model
V's daily excess return regressed on Fama/French 5 factors plus momentum (Ken French library, which ends 2026-08-31). Newey-West t-stats. Rolling 252-day betas show how the exposures drift.

```python
import warnings

warnings.filterwarnings("ignore")
import matplotlib.pyplot as plt

from pitch import analysis as a
from pitch.plotting import apply_theme, save

apply_theme()
```

```python
fm, rolling = a.factor_model()
fm
```

```python
fig, ax = plt.subplots(figsize=(9, 4))
rolling.plot(ax=ax, lw=1.2)
ax.axhline(0, color="#374151", lw=0.8)
ax.set_title("V rolling 252-day factor betas")
ax.legend(ncol=6, fontsize=8, frameon=False)
save(fig, "02_rolling_betas")
```
