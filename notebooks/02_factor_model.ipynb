# Factor model

Daily excess return of UNH on the Fama/French five factors plus momentum, with Newey-West t-stats. The Ken French files end 2026-08-31. The same code writes the rolling-beta chart for COST and for archived V under `figures/<TICKER>/`.


```python
import warnings

warnings.filterwarnings("ignore")
import matplotlib.pyplot as plt

from pitch import analysis as a
from pitch.plotting import apply_theme, save
from pitch.universe import ACTIVE

apply_theme()
TICKERS = [ACTIVE, "COST", "V"]  # active pitch, comparison, archived first pick
a.set_pitch(ACTIVE)
```

```python
fm, rolling = a.factor_model()
fm
```

```python
for t in TICKERS:
    a.set_pitch(t)
    _, rolling = a.factor_model()
    fig, ax = plt.subplots(figsize=(9, 4))
    rolling.plot(ax=ax, lw=1.2)
    ax.axhline(0, color="#374151", lw=0.8)
    ax.set_title(f"{t} rolling 252-day factor betas")
    ax.legend(ncol=6, fontsize=8, frameon=False)
    save(fig, f"{t}/02_rolling_betas")
a.set_pitch(ACTIVE)
```
