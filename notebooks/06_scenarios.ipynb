# 06 · Scenarios
Runs for the active pitch (**UNH**), with tables shown for it. It also redraws the figures for COST (comparison) and V (archived first pick), using the same code and assumptions. Figures go to `figures/<TICKER>/`.

Bear, base and bull cases reuse the DCF. Stage-1 growth is set at the 25th, 50th and 75th percentiles of the stock's own rolling 5-year FCF growth rates, and WACC moves by ±0.5pp.

```python
import warnings

warnings.filterwarnings("ignore")
import matplotlib.pyplot as plt

from pitch import analysis as a
from pitch.plotting import apply_theme, save
from pitch.universe import ACTIVE

apply_theme()
TICKERS = [ACTIVE, "COST", "V"]  # active pitch, comparison, archived first pick
a.set_pitch(ACTIVE)
```

```python
a.scenarios()
```

```python
names = ["bear", "base", "bull"]
for t in TICKERS:
    a.set_pitch(t)
    s = a.scenarios()
    fig, ax = plt.subplots(figsize=(7, 3.6))
    ax.bar(names, [s[n]["target"] for n in names], color=["#f87171", "#94a3b8", "#10B981"])
    ax.axhline(a.dcf()["price"], color="#f59e0b", ls="--", label="as-of price")
    ax.set_title(t)
    ax.set_ylabel("target per share")
    ax.legend(frameon=False)
    save(fig, f"{t}/06_scenarios")
a.set_pitch(ACTIVE)
```
