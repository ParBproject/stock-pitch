# 06 · Scenarios
Bear, base and bull cases reuse the DCF. Stage-1 growth is set at the 25th, 50th and 75th percentiles of V's rolling 5-year FCF growth rates, and WACC moves by ±0.5pp.

```python
import warnings

warnings.filterwarnings("ignore")
import matplotlib.pyplot as plt

from pitch import analysis as a
from pitch.plotting import apply_theme, save

apply_theme()
```

```python
s = a.scenarios()
s
```

```python
names = ["bear", "base", "bull"]
fig, ax = plt.subplots(figsize=(7, 3.6))
ax.bar(names, [s[n]["target"] for n in names], color=["#f87171", "#94a3b8", "#10B981"])
ax.axhline(a.dcf()["price"], color="#f59e0b", ls="--", label="as-of price")
ax.set_ylabel("target per share")
ax.legend(frameon=False)
save(fig, "06_scenarios")
```
