# Comps

EV/EBITDA, P/E and FCF yield from the latest 10-K and the as-of price. Market cap is the as-of price times the Yahoo share count. Missing XBRL items, losses and lender rows are listed in `gaps` and left blank. The same code writes the COST and V charts.


```python
import warnings

warnings.filterwarnings("ignore")
import matplotlib.pyplot as plt

from pitch import analysis as a
from pitch.plotting import apply_theme, save
from pitch.universe import ACTIVE

apply_theme()
TICKERS = [ACTIVE, "COST", "V"]  # active pitch, comparison, archived first pick
a.set_pitch(ACTIVE)
```

```python
c = a.comps()
c[["fiscal_year_end", "price", "market_cap", "ev_ebitda", "pe", "fcf_yield", "gaps"]]
```

```python
for t in TICKERS:
    a.set_pitch(t)
    c = a.comps()
    fig, axes = plt.subplots(1, 3, figsize=(11, 3.6))
    for ax, col in zip(axes, ["ev_ebitda", "pe", "fcf_yield"], strict=True):
        s = c[col].dropna().sort_values()
        ax.barh(s.index, s.values, color=["#10B981" if x == t else "#374151" for x in s.index])
        ax.set_title(col)
    save(fig, f"{t}/03_comps")
a.set_pitch(ACTIVE)
```
