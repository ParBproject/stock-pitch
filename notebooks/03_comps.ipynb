# 03 · Comps
Runs for the active pitch (**UNH**), with tables shown for it. It also redraws the figures for COST (comparison) and V (archived first pick), using the same code and assumptions. Figures go to `figures/<TICKER>/`.

EV/EBITDA, P/E and FCF yield come from the latest 10-K (SEC XBRL) and the as-of price. Market cap = as-of price x Yahoo share count.

**Gaps** are stated per row in the `gaps` column and never filled:
- figures not reported in XBRL
- net losses or negative EBITDA, where the multiple isn't meaningful
- lenders, for V's peer set

```python
import warnings

warnings.filterwarnings("ignore")
import matplotlib.pyplot as plt

from pitch import analysis as a
from pitch.plotting import apply_theme, save
from pitch.universe import ACTIVE

apply_theme()
TICKERS = [ACTIVE, "COST", "V"]  # active pitch, comparison, archived first pick
a.set_pitch(ACTIVE)
```

```python
c = a.comps()
c[["fiscal_year_end", "price", "market_cap", "ev_ebitda", "pe", "fcf_yield", "gaps"]]
```

```python
for t in TICKERS:
    a.set_pitch(t)
    c = a.comps()
    fig, axes = plt.subplots(1, 3, figsize=(11, 3.6))
    for ax, col in zip(axes, ["ev_ebitda", "pe", "fcf_yield"], strict=True):
        s = c[col].dropna().sort_values()
        ax.barh(s.index, s.values, color=["#10B981" if x == t else "#374151" for x in s.index])
        ax.set_title(col)
    save(fig, f"{t}/03_comps")
a.set_pitch(ACTIVE)
```
