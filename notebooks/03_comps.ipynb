# 03 · Comps
EV/EBITDA, P/E and FCF yield from the latest 10-K (SEC XBRL) and the as-of price. Market cap = as-of price x Yahoo share count.

**Gaps:** EV/EBITDA and FCF yield are not meaningful for lenders (AXP, COF, SYF, ALLY), so they are left blank. Visa's share count comes from Yahoo and may not include its class B/C shares on an as-converted basis.

```python
import warnings

warnings.filterwarnings("ignore")
import matplotlib.pyplot as plt

from pitch import analysis as a
from pitch.plotting import apply_theme, save

apply_theme()
```

```python
c = a.comps()
c[["fiscal_year_end", "price", "market_cap", "ev_ebitda", "pe", "fcf_yield", "gaps"]]
```

```python
fig, axes = plt.subplots(1, 3, figsize=(11, 3.6))
for ax, col in zip(axes, ["ev_ebitda", "pe", "fcf_yield"], strict=True):
    s = c[col].dropna().sort_values()
    ax.barh(s.index, s.values, color=["#10B981" if t == "V" else "#374151" for t in s.index])
    ax.set_title(col)
save(fig, "03_comps")
```
