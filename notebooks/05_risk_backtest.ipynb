# 05 · Risk and walk-forward backtest
Runs for the active pitch (**UNH**), with tables shown for it. It also redraws the figures for COST (comparison) and V (archived first pick), using the same code and assumptions. Figures go to `figures/<TICKER>/`.

Risk metrics cover the full sample and the trailing 3 years. The backtest uses a trend filter whose window is re-chosen each year from the trailing 3 years only, trades the next day, and pays 10 bp per switch. It is compared with buy-and-hold and SPY over the same out-of-sample period.

```python
import warnings

warnings.filterwarnings("ignore")
import matplotlib.pyplot as plt

from pitch import analysis as a
from pitch.plotting import apply_theme, save
from pitch.universe import ACTIVE

apply_theme()
TICKERS = [ACTIVE, "COST", "V"]  # active pitch, comparison, archived first pick
a.set_pitch(ACTIVE)
```

```python
a.risk()
```

```python
bt, _ = a.backtest()
bt
```

```python
for t in TICKERS:
    a.set_pitch(t)
    _, curves = a.backtest()
    fig, ax = plt.subplots(figsize=(9, 4))
    curves.rename(columns={"buy_hold": f"buy-and-hold {t}"}).plot(ax=ax, lw=1.3)
    ax.set_yscale("log")
    ax.set_title(f"{t}: growth of $1, out-of-sample (net of costs)")
    ax.legend(frameon=False)
    save(fig, f"{t}/05_backtest")
a.set_pitch(ACTIVE)
```
