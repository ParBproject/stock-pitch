# 05 · Risk and walk-forward backtest
Risk metrics cover the full sample and the trailing 3 years. The backtest uses a trend filter on V whose window is re-chosen each year from the trailing 3 years only, trades the next day, and pays 10 bp per switch. It is compared with buy-and-hold V and SPY over the same out-of-sample period.

```python
import warnings

warnings.filterwarnings("ignore")
import matplotlib.pyplot as plt

from pitch import analysis as a
from pitch.plotting import apply_theme, save

apply_theme()
```

```python
a.risk()
```

```python
bt, curves = a.backtest()
bt
```

```python
fig, ax = plt.subplots(figsize=(9, 4))
curves.plot(ax=ax, lw=1.3)
ax.set_yscale("log")
ax.set_title("Growth of $1, out-of-sample (net of costs)")
ax.legend(frameon=False)
save(fig, "05_backtest")
```
